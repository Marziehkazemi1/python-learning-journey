```python
import seaborn as sns

df = sns.load_dataset("penguins")
df
```

```python
sns.histplot(
    data=df,
    x="body_mass_g",
    kde=True,
    bins=30
)
```

```python
skewness = df['body_mass_g'].skew()
print(skewness)
```

```python
b1 = df.groupby("species")["body_mass_g"].skew()
print(b1)
```
