```python
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("tips")

ax = sns.scatterplot(
    data=df,
    x="total_bill",
    y="tip",
    hue="sex"
)

ax.set_title("Total Bill vs Tip")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")
```
