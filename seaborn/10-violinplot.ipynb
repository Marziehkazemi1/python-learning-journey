```python
import seaborn as sns

df = sns.load_dataset('penguins')
df
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    hue="sex",
)
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    hue="sex",
    split=True
)
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    inner="quart"
)
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    cut=0
)
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    bw_adjust=0.5
)
```

```python
sns.violinplot(
    data=df,
    x="species",
    y="body_mass_g",
    density_norm="count"
)
```
