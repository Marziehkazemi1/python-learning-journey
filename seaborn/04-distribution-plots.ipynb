```python
import seaborn as sns

penguins = sns.load_dataset("penguins")
penguins
```

```python
sns.histplot(penguins, x='body_mass_g', bins=20, hue='species', kde=True)
```

```python
sns.kdeplot(penguins, x='body_mass_g', fill=True, hue='sex', bw_adjust=0.5)
```

```python
sns.displot(penguins, x='body_mass_g', kind='hist', bins=20)
```

```python
sns.displot(penguins, x='body_mass_g', kind='ecdf')
```

```python
sns.displot(penguins, x='body_mass_g', kind='kde', col='species')
```
